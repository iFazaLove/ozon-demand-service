```python
import os
import mlflow
from dotenv import load_dotenv

load_dotenv()
```

```python
tracking_uri = os.getenv("MLFLOW_TRACKING_URI", "http://127.0.0.1:5000")
mlflow.set_tracking_uri(tracking_uri)
mlflow.set_experiment("test-testovich")

history = [6, 8, 7, 5, 9, 10, 11]
actual = 8

forecasts = {
    "last_week": history[0],
    "weekly_mean": sum(history) / len(history),
}

for method, forecast in forecasts.items():
    with mlflow.start_run(run_name=method) as run:
        mlflow.log_params({
                "method": method,
                "dataset": "synthetic",
                "horizon_days": 1,
        })
        mlflow.log_metrics({
            "mae_units": abs(forecast - actual)
        })
        mlflow.log_dict(
            {
                "actual": actual,
                "forecast": forecast,
            },
            "forecast.json",
        )
        print(f"{method}: run_id={run.info.run_id}")
```
